```python
import pandas as pd
import pickle
from src.recommender import MovieRecommender
```

```python
movies = pd.read_csv("../data/processed/imdb_preprocessed.csv")

movies.head()
```

```python
with open("../models/tfidf_matrix.pkl", "rb") as file:
    tfidf_matrix = pickle.load(file)
```

```python

recommender = MovieRecommender(
    movies,
    tfidf_matrix
)
```

```python
user_mood = input("How are you feeling today? ")
favourite_movie = input("Enter your favourite movie: ")
```

```python
result = recommender.recommend(
    mood_input=user_mood,
    favourite_movie=favourite_movie,
    top_n=10
)
```

```python
print("Predicted Mood:", result["predicted_mood"])
print("Matched Movie:", result["matched_movie"])

display(result["recommendations"])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
movies = pd.read_csv("../data/raw/IMDB-Movie-Dataset(2023-1951).csv")
```

## Q1. Which year produced the most movies ?

```python
movies["year"].value_counts().sort_index().plot(figsize=(12,5))

plt.title("Movies Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")

plt.show()
movies_per_year = movies['year'].value_counts().sort_index()
print(movies_per_year)
```

```python
max_year = movies_per_year.idxmax()
max_movies = movies_per_year.max()

print(f"Year with maximum movie releases: {max_year}")
print(f"Number of movies released: {max_movies}")
```

```python
top_10 = movies['year'].value_counts().head(20)
print(top_10)
import matplotlib.pyplot as plt

movies_per_year.plot(kind='bar', figsize=(15, 6))
plt.title("Number of Movies Released Per Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.xticks(rotation=90)
plt.show()
```

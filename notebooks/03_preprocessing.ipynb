```python
import pandas as pd

movies = pd.read_csv("../data/raw/IMDB-Movie-Dataset(2023-1951).csv")
```

```python
movies.isnull().sum()
```

```python
movies["year"] = movies["year"].fillna("Unknown")
```

```python
movies = movies[
    ["movie_name",
     "year",
     "genre",
     "overview",
     "director",
     "cast"
     ]
]
```

```python
movies.head()
```

```python
movies["movie_name"] = movies["movie_name"].str.lower()
movies["genre"] = movies["genre"].str.lower()
movies["overview"] = movies["overview"].str.lower()
movies["director"] = movies["director"].str.lower()
movies["cast"] = movies["cast"].str.lower()
```

```python
movies["movie_name"] = movies["movie_name"].str.strip()
movies["director"] = movies["director"].str.strip()
movies["cast"] = movies["cast"].str.strip()
movies["genre"] = movies["genre"].str.strip()
```

```python
movies["genre"] = movies["genre"].str.replace(", ", " ")
movies["cast"] = movies["cast"].str.replace(", ", " ")
```

```python
movies.head()
```

```python
movies["tags"] = (
    movies["genre"] + " " +
    movies["overview"] + " " +
    movies["director"] + " " +
    movies["cast"] + " " 
)
```

```python
new_df = movies[
    ["movie_name", "year", "tags"]
]
print(new_df.head())
```

```python
new_df["tags"] = new_df["tags"].str.lower()
```

```python
new_df.to_csv(
    "../data/processed/imdb_cleaned.csv",
    index=False
)
```

```python
import nltk
import string

from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem.wordnet import WordNetLemmatizer
```

```python
nltk.download("stopwords")
```

```python
ps = PorterStemmer()
```

```python
def preprocess(text):

    # convert to lowercase
    text = text.lower()

    # remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # split into words
    words = text.split()

    # remove stopwords
    words = [word for word in words if word not in stopwords.words("english")]

    # stemming
    words = [ps.stem(word) for word in words]

    return " ".join(words)
```

```python
new_df["tags"] = new_df["tags"].apply(preprocess)
```

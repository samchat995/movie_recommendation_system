```python
import pandas as pd

movies = pd.read_csv("../data/processed/imdb_preprocessed.csv")

movies.head()
```

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.append(str(project_root))
```

```python
from src.movie_filter import MovieFilter
from src.mood_classifier import MoodClassifier
from src.genre_mapper import MOOD_GENRES
```

```python
user_input = "I had a stressful day."
```

```python
classifier = MoodClassifier()

mood = classifier.predict(user_input)

print(mood)
```

```python
genres = MOOD_GENRES[mood]
print(genres)
```

```python
movie_filter = MovieFilter(movies)
print(movie_filter)

filtered_movies = movie_filter.filter_by_genres(genres)
filtered_movies.head()
```

```python
candidate_indices = filtered_movies.index.tolist()

print(candidate_indices[:10])
```
